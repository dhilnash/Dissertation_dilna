# **01_Data_Load.py**


Objective: Before data cleansing, examine and confirm the raw datasets.

Inputs:
The raw_data folder contains the following raw datasets:

*   Elliptic_txs_features.csv
*  Elliptic_txs_classes.csv
*  Transaction_dataset.csv




Inspections performed: The code verifies the representation of each row in the dataset, the dimensions and layout of the dataset, the column names, and the kind of data contained in each column. It additionally examines if any data is absent, if there are repeated entries, and if the target classes and categories are evenly distributed.

Reasults: The code reveals the dimensions and column details of every dataset, overviews of the data types, counts of missing entries, counts of duplicate records, and the distribution of target categories and classes.

```python
import os
import pandas as pd
```

```python
raw_data_path = "../Raw_data"

print("Files in Raw_data:")

for file in os.listdir(raw_data_path):
    print(file)
```

```python
pd.set_option("display.max_columns", 10)
pd.set_option("display.width", 120)
```

```python
def inspect(df, name, target_col=None):
    print(f"\n{'='*60}\n{name}\n{'='*60}")
    print(f"Shape: {df.shape[0]} rows, {df.shape[1]} columns")
    print(f"\nColumn dtypes:\n{df.dtypes.value_counts()}")
    print(f"\nMissing values (top 5 columns):\n{df.isna().sum().sort_values(ascending=False).head(5)}")
    print(f"\nTotal missing values: {df.isna().sum().sum()}")
    print(f"Duplicate rows: {df.duplicated().sum()}")
    if target_col:
        print(f"\nTarget column '{target_col}' value counts:\n{df[target_col].value_counts()}")
```

```python
# ELLIPTIC: one row represents one Bitcoin TRANSACTION

print("ELLIPTIC: each row = one transaction. Target: 'class' column,")
print("coded '1' = illicit, '2' = licit, 'unknown' = unlabelled (verified")
print("against the dataset's own documentation, not assumed).")
```

```python
ell_features = pd.read_csv(
    "../Raw_data/elliptic_txs_features.csv",
    header=None
)
inspect(ell_features, "Elliptic features (raw, no header)")
```

```python
ell_classes = pd.read_csv("../Raw_data/elliptic_txs_classes.csv")
inspect(ell_classes, "Elliptic classes", target_col="class")
```

```python
# ETHEREUM: one row represents one ACCOUNT (not a transaction)

print("\n\nETHEREUM: each row = one account (aggregated transaction")
print("history), not a single transaction. Target: 'FLAG' column,")
print("coded 0 = legitimate, 1 = fraudulent.")
```

```python
eth = pd.read_csv("../Raw_data/transaction_dataset.csv")
inspect(eth, "Ethereum transaction_dataset.csv", target_col="FLAG")
```

```python
print("\n\nSummary: both datasets verified readable, target columns confirmed,")
print("and initial duplicate/missing-value counts captured here as a baseline")
print("to compare against after cleaning in 02_Data_Cleaning.py.")

```
