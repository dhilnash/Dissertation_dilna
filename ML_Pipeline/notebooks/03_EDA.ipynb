# **03_EDA.py**

Purpose:
Conduct exploratory data analysis (EDA) on cleansed Elliptic and Ethereum cryptocurrency datasets before machine learning modeling.

Inputs:
Cleaned data files:

*   elliptic_clean.csv
*   ethereum_clean.csv



Outputs:
EDA tables and visualizations highlight class distribution, dataset structure, feature distributions, and significant patterns.


*    eda_charts/A_elliptic_class_balance.png


*   eda_charts/B_elliptic_illicit_rate_by_time.png

*   eda_charts/C_elliptic_top_features.png



*    eda_charts/D_ethereum_class_balance.png
*   eda_charts/E_ethereum_feature_boxplots.png


*  eda_charts/F_ethereum_correlation.png





```python
import os
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt


```

```python
print(os.listdir("../ML_Pipeline"))
```

```python
print(os.listdir("../ML_Pipeline/cleaned_data"))
```

```python
NAVY, TEAL, RED = "#1E3A5F", "#0EA5A6", "#DC2626"
plt.rcParams["font.family"] = "DejaVu Sans"

os.makedirs("eda_charts", exist_ok=True)
```

```python
ell = pd.read_csv(
    "../cleaned_data/elliptic_clean.csv"
)

eth = pd.read_csv(
    "../cleaned_data/ethereum_clean.csv"
)

print("Elliptic shape:", ell.shape)
print("Ethereum shape:", eth.shape)
```

```python
eda_charts_path = "../ML_Pipeline/eda_charts"
os.makedirs(eda_charts_path, exist_ok=True)

print("EDA charts folder ready:")
print(eda_charts_path)
```

```python
# A: Elliptic class balance

fig, ax = plt.subplots(figsize=(5, 4), dpi=150)

counts = ell["label"].value_counts().sort_index()

ax.bar(
    ["Licit", "Illicit"],
    counts.values,
    color=[TEAL, RED]
)

ax.set_title("Elliptic: Class Balance (Full Labelled Set)")
ax.set_ylabel("Number of Transactions")

fig.tight_layout()

fig.savefig(
    os.path.join(eda_charts_path, "A_elliptic_class_balance.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)
```

```python
print(ell["label"].value_counts())
print(ell["label"].value_counts(normalize=True) * 100)
```

```python
# B: Elliptic illicit rate by time step

by_step = ell.groupby("feat_1")["label"].mean()

fig, ax = plt.subplots(figsize=(9, 4), dpi=150)

ax.plot(
    by_step.index,
    by_step.values,
    color=NAVY,
    marker="o",
    markersize=3
)

ax.set_xlabel("Time Step")
ax.set_ylabel("Illicit Rate")
ax.set_title("Elliptic: Illicit Rate by Time Step")

fig.tight_layout()

fig.savefig(
    os.path.join(eda_charts_path, "B_elliptic_illicit_rate_by_time.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)
```

```python
# C: Elliptic top features by standardized mean difference

feat_cols = [
    c for c in ell.columns
    if c.startswith("feat_") and c != "feat_1"
]

illicit = ell[ell["label"] == 1][feat_cols]
licit = ell[ell["label"] == 0][feat_cols]

diffs = {
    c: (illicit[c].mean() - licit[c].mean()) / ell[c].std()
    for c in feat_cols
    if ell[c].std() > 0
}

top = (
    pd.Series(diffs)
    .sort_values(key=abs, ascending=False)
    .head(15)
)

fig, ax = plt.subplots(figsize=(8, 5.5), dpi=150)

ax.barh(
    top.index[::-1],
    top.values[::-1],
    color=[
        RED if v > 0 else TEAL
        for v in top.values[::-1]
    ]
)

ax.set_title("Elliptic: Top 15 Features by Illicit/Licit Separation")
ax.set_xlabel("Standardized Mean Difference")

fig.tight_layout()

fig.savefig(
    os.path.join(
        eda_charts_path,
        "C_elliptic_top_features.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)
```

```python
# D: Ethereum class balance

fig, ax = plt.subplots(figsize=(5, 4), dpi=150)

counts = eth["FLAG"].value_counts().sort_index()

ax.bar(
    ["Legit", "Fraud"],
    counts.values,
    color=[TEAL, RED]
)

ax.set_title("Ethereum: Class Balance")
ax.set_ylabel("Number of Addresses")

fig.tight_layout()

fig.savefig(
    os.path.join(
        eda_charts_path,
        "D_ethereum_class_balance.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)
```

```python
print(eth["FLAG"].value_counts())

print(eth["FLAG"].value_counts(normalize=True) * 100)
```

```python
# E: Ethereum feature boxplots by class

features = [
    "Sent tnx",
    "Received Tnx",
    "total ether balance",
    "Time Diff between first and last (Mins)"
]

fig, axes = plt.subplots(
    1, 4,
    figsize=(14, 4),
    dpi=150
)

for ax, feat in zip(axes, features):

    bp = ax.boxplot(
        [
            eth[eth["FLAG"] == 0][feat],
            eth[eth["FLAG"] == 1][feat]
        ],
        tick_labels=["Legit", "Fraud"],
        patch_artist=True,
        showfliers=False
    )

    for patch, color in zip(
        bp["boxes"],
        [TEAL, RED]
    ):
        patch.set_facecolor(color)
        patch.set_alpha(0.7)

    ax.set_title(feat, fontsize=9)

fig.tight_layout()

fig.savefig(
    os.path.join(
        eda_charts_path,
        "E_ethereum_feature_boxplots.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)
```

```python
# F: Ethereum correlation with FLAG

numeric_cols = [
    c for c in eth.select_dtypes(include=[np.number]).columns
    if c != "FLAG"
]

corrs = (
    eth[numeric_cols]
    .corrwith(eth["FLAG"])
    .sort_values(key=abs, ascending=False)
    .head(15)
)

fig, ax = plt.subplots(figsize=(8, 5.5), dpi=150)

ax.barh(
    corrs.index[::-1],
    corrs.values[::-1],
    color=[
        RED if v > 0 else TEAL
        for v in corrs.values[::-1]
    ]
)

ax.set_title("Ethereum: Top 15 Features Correlated with Fraud Label")
ax.set_xlabel("Correlation with Fraud Label")

fig.tight_layout()

fig.savefig(
    os.path.join(
        eda_charts_path,
        "F_ethereum_correlation.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)
```
